## Exercise 1: The birthday problem

**Question.** What is the probability that at least two people in this room share a birthday?

More precisely: let $B_n$ be the event that at least two people out of $n$ share a birthday. Find $\Pr(B_n)$.

**Assumptions:**

```python
# Your code here: implement Pr(B_n) and evaluate it for n = 17 and n = 23
def pr_bn(n, days = 365):
    probs = []
    n = int(n)
    for i in range(n):
        probs.append((days-i)/days)
    pr = np.prod(probs)
    return 1 - pr

print(pr_bn(17))
print(pr_bn(23))
print('23 people give roughly even odds')
```

```python
# Your code here: plot Pr(B_n) as a function of n
n = np.linspace(1, 60, 60)
probs = []
for i in n:
    probs.append(pr_bn(i))

plt.plot(n, probs)
plt.xlabel('n - number of people')
plt.ylabel('prob of people sharing birthday')
plt.axhline(0.5, color = 'red')
plt.axvline(23, color = 'red')
plt.show()
```
